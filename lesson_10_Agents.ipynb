```python
@tool
def product_query(query:str, max_price: int | None = None)-> list[dict]:
    """Search available products using a product query.
    The query does not need to exactly match the product name.
    An optional max_price can be provided to exclude products above that price."""
    
    products = [
        {"id": 1, "name": "monitor", "price": 300},
        {"id": 2, "name": "gpu", "price": 500},
        {"id": 3, "name": "cpu", "price": 300},
    ]

    return [prod for prod in products if query.lower() in prod["name"].lower() and (max_price is None or prod["price"] <= max_price)]
```

```python
@tool
def get_product_details(product_id: int) -> dict:
    """Get a details of available products using product id(id, name, manufacture, price )"""

    products = [
        {"id": 1, "name": "Monitor", "manufacture":"ASUS" , "price": 300},
        {"id": 2, "name": "GPU","manufacture":"Nvidia" ,"price": 500},
        {"id": 3, "name": "CPU", "manufacture":"Intel", "price": 300},
    ]

    for product_info in products:
        if product_info["id"] == product_id:
            return product_info
    raise ValueError(f"Invalid product ID: {product_id}")
```

```python
# Build the agent

agent = create_agent(
    model= llm,
    tools=[product_query,get_product_details]
    )
```

```python
result = agent.invoke({
    "messages":[{
        "role":"user",
        "content": "Tell me the details of any GPU under $600."
    }]
    })
```

```python
print(result)
```

```python
for message in result["messages"]:
    print(f"{message.__class__.__name__}:")
    print(message)
    print("-" * 50)
```
